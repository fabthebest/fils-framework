## Part 1 — Variables & Types

A variable is a **named container** that holds a value. The name is yours to choose. The value can change. Python figures out the *type* automatically — you never have to declare it.

There are four types you will use constantly:
- `int` — whole numbers (`42`, `-7`)
- `float` — decimal numbers (`3.14`, `-0.5`)
- `str` — text, always in quotes (`"hello"`, `'world'`)
- `bool` — truth values (`True`, `False`)

> **Fabrice the Skeptic says:** *"Why do types matter? Can't Python just figure it out?"*
> Python does figure out the type — but YOU need to know the type, because `"3" + "4"` gives you `"34"`, not `7`. The machine is consistent. You have to be conscious.

```python
# ============================================================
# VARIABLES & TYPES — Augmented Cell 1.1
# ============================================================

# [CORE] A variable is a label attached to a value stored in memory.
#        The = sign does NOT mean 'equals' — it means 'assign'.
#        Right side is evaluated first, then stored under the left-side name.
restaurant_name = "Le Palais"
table_count = 24
average_bill = 47.50
is_open = True

# [ORBIT: Restaurant] The variable name is the LABEL on a storage container
#                     in the kitchen stockroom. 'restaurant_name' is a jar
#                     marked 'Name'. What's inside the jar can change —
#                     tomorrow it might say "Le Bistro" — but the jar's
#                     label stays the same until you rename it.

# [ORBIT: Construction] A variable is a slot on the blueprint's data sheet.
#                       'table_count = 24' means the blueprint now officially
#                       records: this building has 24 tables. Any foreman
#                       reading the sheet gets that number immediately.

# [MIRROR] In real life, a label IS permanently tied to its container.
#          You can't take the 'sugar' label and slap it on the salt jar.
#          In Python you absolutely can — variable names are just pointers.
#          'restaurant_name = 42' is legal. Python won't object.

# [TRAP] Never use spaces in variable names. 'restaurant name' causes a
#        SyntaxError. Use underscores: 'restaurant_name'. Also, Python
#        is case-sensitive: 'Table_Count' and 'table_count' are different variables.

print(restaurant_name)
print(table_count)
print(average_bill)
print(is_open)
```

```python
# ============================================================
# VARIABLES & TYPES — Augmented Cell 1.2  (type inspection)
# ============================================================

# [CORE] type() is a built-in function that returns the type of any value.
#        This is how you interrogate Python: "what ARE you?"
print(type(restaurant_name))   # <class 'str'>
print(type(table_count))       # <class 'int'>
print(type(average_bill))      # <class 'float'>
print(type(is_open))           # <class 'bool'>

# [ORBIT: Restaurant] The health inspector checks each storage container
#                     and stamps its contents: LIQUID, SOLID, DRY-GOODS.
#                     type() is Python's health inspector. It doesn't
#                     change the contents — it just tells you what's there.

# [ORBIT: Construction] The site surveyor reads the blueprint data sheet
#                       and confirms each entry is in the right column:
#                       INTEGER column, DECIMAL column, TEXT column.
#                       Putting a measurement in the wrong column
#                       causes calculation errors downstream.

# [MIRROR] In a kitchen, the health inspector can FORCE you to relabel
#          a container if it's mislabeled. Python's type() cannot change
#          the type — it only reports it. To change the type you must
#          explicitly convert: int(), str(), float().

# [TRAP] The string "42" and the integer 42 look the same to human eyes
#        but are entirely different to Python.
age_as_string = "42"
age_as_int = 42
print(age_as_string + " years")   # Works: string concatenation
print(age_as_int + 8)             # Works: integer addition
# print(age_as_string + 8)        # CRASHES: can't add str and int — uncomment to see
```

```python
# ============================================================
# VARIABLES & TYPES — Augmented Cell 1.3  (type conversion)
# ============================================================

# [CORE] Type conversion (casting) lets you change a value's type.
#        int(), float(), str(), bool() are the main conversion functions.
#        They do NOT modify the original variable — they return a NEW value.
raw_input = "15"            # imagine this came from a user typing in a form
guests = int(raw_input)     # now it's a number we can do math with
bill_per_person = average_bill / guests

print(f"Guests: {guests}")
print(f"Bill per person: ${bill_per_person:.2f}")

# [ORBIT: Restaurant] The front-of-house manager takes a handwritten note
#                     ('15' — written text) and translates it into a number
#                     on the kitchen whiteboard so the chefs can count
#                     plates. The note doesn't change. A new number appears
#                     on the board. Two representations, one reality.

# [ORBIT: Construction] The architect's sketch says 'fifteen' in cursive.
#                       The structural engineer converts that to the integer 15
#                       before plugging it into the load-bearing formula.
#                       The sketch stays unchanged in the folder.

# [MIRROR] In real life, converting a handwritten '15' to a number on a
#          whiteboard is irreversible once erased. In Python, raw_input
#          still holds '15' after the conversion. int() made a copy,
#          not a replacement.

# [TRAP] int('3.7') will CRASH — you cannot convert a decimal string
#        directly to int. You must go through float first: int(float('3.7'))
print(int(float("3.7")))    # works: gives 3 (truncates, does not round)
```

## Part 2 — Lists & Dictionaries

Single variables hold one value. But the real world is full of *collections*. Python gives you two essential collection types:

- **List** `[]` — an ordered sequence. Position matters. Item 0 is always the first.
- **Dictionary** `{}` — a key-value map. No fixed order. You look things up by name, not by position.

The choice between them is architectural. A list answers *"what's third?"* A dictionary answers *"what's the price of salmon?"*

```python
# ============================================================
# LISTS — Augmented Cell 2.1
# ============================================================

# [CORE] A list is an ordered, mutable (changeable) collection.
#        Items are indexed starting at 0. Negative indices count from the end.
#        Lists can hold items of mixed types, including other lists.
menu_items = ["soup", "salad", "steak", "salmon", "sorbet"]

print(menu_items[0])     # 'soup'  — first item
print(menu_items[-1])    # 'sorbet' — last item (negative indexing)
print(menu_items[1:3])   # ['salad', 'steak'] — slice: index 1 up to (not including) 3
print(len(menu_items))   # 5 — total number of items

# Modifying a list
menu_items.append("soufflé")       # add to the end
menu_items.insert(1, "bruschetta") # insert at position 1, shifts everything right
menu_items.remove("salad")         # remove first occurrence of this value

print(menu_items)

# [ORBIT: Restaurant] A list is the ORDER TICKET on a kitchen spike.
#                     Items come in sequence — course 1, course 2, course 3.
#                     Position is everything. The kitchen fires course 0 first,
#                     course 1 second. You can add a last-minute dish (append)
#                     or squeeze in an amuse-bouche before the main (insert).

# [ORBIT: Construction] A list is the TASK SEQUENCE on a Gantt chart.
#                       Pour foundation [0], frame walls [1], run electrical [2].
#                       You cannot skip index 0. The order is the contract.

# [MIRROR] An order ticket at a restaurant cannot easily have an item
#          inserted in the middle once it's written — a waiter would rewrite
#          the whole thing. Python's insert() does this instantly with no cost
#          visible to you, but internally it shifts every subsequent item.
#          On large lists, insert(0, x) is slow — it touches every element.

# [TRAP] Indexing starts at 0, not 1. menu_items[5] on a 5-item list
#        causes an IndexError. The last valid index is always len(list) - 1.
```

```python
# ============================================================
# DICTIONARIES — Augmented Cell 2.2
# ============================================================

# [CORE] A dictionary maps keys to values. Keys must be unique and immutable
#        (strings and numbers work; lists do not). Values can be anything.
#        Lookup by key is extremely fast regardless of dictionary size.
dish_prices = {
    "soup":    8.00,
    "steak":  34.00,
    "salmon": 28.50,
    "sorbet":  9.00
}

print(dish_prices["steak"])           # 34.0
print(dish_prices.get("pasta", 0.0))  # 0.0 — safe lookup, no crash if missing

# Adding and updating
dish_prices["truffle pasta"] = 42.00  # new key: adds it
dish_prices["steak"] = 36.00          # existing key: overwrites it

# Iterating
for dish, price in dish_prices.items():
    print(f"  {dish:20s} ${price:.2f}")

# [ORBIT: Restaurant] A dictionary is the PRICE BOARD behind the bar.
#                     You don't ask 'what's item number 3?' — you ask
#                     'what does salmon cost?' The board looks up by NAME.
#                     Each drink has exactly one price. Duplicating a key
#                     would be like listing salmon twice — only the last
#                     entry survives.

# [ORBIT: Construction] A dictionary is the MATERIALS COST SHEET.
#                       Key = material name, Value = unit cost.
#                       'concrete': 120.00, 'steel': 890.00.
#                       The estimator looks up 'steel' directly — no scanning
#                       through a numbered list.

# [MIRROR] A price board is physically fixed in one spot. Python dicts
#          have no inherent location — they live in memory. Also, a real
#          price board CAN list the same item twice (accidental). A dict
#          silently overwrites the old price. No warning. No error.

# [TRAP] dish_prices["lobster"] on a key that doesn't exist raises a KeyError.
#        Always use .get(key, default) when the key might be absent.
```

## Part 3 — Loops & Conditions

Loops and conditions are how code makes decisions and repeats work. Together they give your program *will* — it can adapt to what it finds.

- `if / elif / else` — branching: do this *or* that based on a condition
- `for` loop — iterate over a known collection
- `while` loop — keep going until a condition is no longer true

> **Fabrice the Skeptic says:** *"Can't I just copy-paste the same line 50 times instead of looping?"*  
> You could — until the requirement changes and you have to edit 50 places instead of 1. Loops are about *maintainability*, not just laziness.

```python
# ============================================================
# CONDITIONS — Augmented Cell 3.1
# ============================================================

# [CORE] An if-statement evaluates a boolean expression.
#        Only one branch executes per run. elif and else are optional.
#        Indentation (4 spaces) defines which code belongs to which branch.
table_occupancy = 18
total_capacity = 24
occupancy_rate = table_occupancy / total_capacity

if occupancy_rate >= 0.90:
    status = "Full house — stop accepting walk-ins"
elif occupancy_rate >= 0.60:
    status = "Busy — priority seating only"
elif occupancy_rate >= 0.30:
    status = "Moderate — open floor"
else:
    status = "Quiet — run lunch specials"

print(f"Occupancy: {occupancy_rate:.0%}  →  {status}")

# [ORBIT: Restaurant] The maitre d' checks occupancy before seating guests.
#                     90%+ full? Hold the door. 60-90%? Regulars only.
#                     Under 30%? Wave the chalkboard special outside.
#                     Each rule is checked in order — first match wins.

# [ORBIT: Construction] The site foreman checks weather before ordering
#                       concrete pours: below 5°C → halt; 5-15°C → add
#                       accelerant; above 15°C → standard mix. One reading,
#                       one decision, no ambiguity.

# [MIRROR] A maitre d' might apply multiple rules simultaneously
#          (a VIP guest AND the restaurant is full → override the rule).
#          Python's if-elif chain stops at the FIRST true branch.
#          For multiple simultaneous conditions, use nested ifs or 'and'/'or'.

# [TRAP] '=' is assignment. '==' is comparison. Writing 'if x = 5:' is a
#        SyntaxError in Python (unlike some other languages where it silently
#        passes). This is one of Python's guardrails — use it.
```

```python
# ============================================================
# FOR LOOPS — Augmented Cell 3.2
# ============================================================

# [CORE] A for loop iterates over any iterable (list, dict, range, string...).
#        The loop variable is rebound on each iteration.
#        range(n) produces integers 0, 1, 2, ... n-1.
daily_covers = [87, 102, 95, 78, 134, 156, 143]  # covers per day, Mon-Sun
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

total = 0
peak_day = days[0]
peak_covers = daily_covers[0]

for i, covers in enumerate(daily_covers):
    total += covers
    if covers > peak_covers:
        peak_covers = covers
        peak_day = days[i]
    print(f"  {days[i]}: {covers:>4} covers")

average = total / len(daily_covers)
print(f"\nWeekly total:   {total}")
print(f"Daily average:  {average:.1f}")
print(f"Peak day:       {peak_day} ({peak_covers} covers)")

# [ORBIT: Restaurant] The restaurant manager reviews the week's reservation
#                     log — one page per day. For each page they tally the
#                     covers, note whether it beat the current record, then
#                     flip to the next page. The loop IS the page-flip.
#                     enumerate() gives you both the page number (i) AND
#                     the page content (covers).

# [ORBIT: Construction] The site inspector walks each floor of a building
#                       (0, 1, 2, ... n-1) ticking off a checklist.
#                       range(7) is the inspector's itinerary: 7 stops.
#                       The clipboard (total) accumulates the findings.

# [MIRROR] A manager physically flips pages in order and cannot skip one
#          without noticing. Python's for loop CAN skip with 'continue'
#          or stop early with 'break'. The loop has levers the analogy lacks.

# [TRAP] Modifying a list WHILE iterating over it causes unpredictable
#        behavior — items get skipped or visited twice. If you need to
#        remove items, iterate over a copy: for item in my_list[:]:
```

## Part 4 — Functions

A function is a **named, reusable block of code**. You define it once, call it many times. Functions can accept inputs (parameters) and return outputs (return values).

The shift from writing code to *writing functions* is the shift from amateur to craftsperson. It is the first act of real software design.

> **Professor Bridge says:** *"If you write the same logic twice, you have already failed the design. Abstract it into a function and call it twice. Now the logic lives in one place, has one name, and can be fixed in one edit."*

```python
# ============================================================
# FUNCTIONS — Augmented Cell 4.1  (definition & calling)
# ============================================================

# [CORE] 'def' defines a function. Parameters are local variables that receive
#        arguments at call time. 'return' sends a value back to the caller.
#        Without 'return', a function implicitly returns None.
def calculate_bill(covers, price_per_cover, tip_percent=18):
    """
    Calculate total bill including tip.

    Parameters
    ----------
    covers : int   — number of guests
    price_per_cover : float — average spend per guest
    tip_percent : float — tip percentage (default 18%)

    Returns
    -------
    dict with subtotal, tip_amount, total
    """
    subtotal = covers * price_per_cover
    tip_amount = subtotal * (tip_percent / 100)
    total = subtotal + tip_amount
    return {
        "subtotal": round(subtotal, 2),
        "tip_amount": round(tip_amount, 2),
        "total": round(total, 2)
    }

# Calling the function — three ways
bill_a = calculate_bill(4, 47.50)            # uses default tip (18%)
bill_b = calculate_bill(4, 47.50, 20)        # explicit tip: 20%
bill_c = calculate_bill(covers=6, price_per_cover=52.00, tip_percent=15)  # keyword args

for label, bill in [("Table A", bill_a), ("Table B", bill_b), ("Table C", bill_c)]:
    print(f"{label}: subtotal=${bill['subtotal']:.2f}  tip=${bill['tip_amount']:.2f}  total=${bill['total']:.2f}")

# [ORBIT: Restaurant] A function is a RECIPE CARD. 'def calculate_bill' is
#                     writing the recipe once and laminating it. Every time
#                     a waiter needs a bill, they hand the recipe card
#                     the specific values (covers, price) — those are the
#                     parameters. The recipe executes and hands back a
#                     finished result. The card itself never changes.

# [ORBIT: Construction] A function is a STANDARD DETAIL DRAWING — a reusable
#                       specification for how to build a staircase, a window
#                       frame, a footing. Every time a foreman needs that
#                       element, they reference the detail number and supply
#                       the site-specific dimensions. One drawing, many uses.

# [MIRROR] A recipe card always produces the same dish given the same
#          ingredients. A Python function CAN produce different results
#          if it reads from global variables or external state. This is
#          a design smell — pure functions (output depends only on inputs)
#          are the ideal.

# [TRAP] Default parameter values are evaluated ONCE at definition time,
#        not at each call. Using a mutable default like def f(items=[]):
#        shares that list across ALL calls — a classic bug. Use None
#        as default and create the mutable object inside the function body.
```

```python
# ============================================================
# FUNCTIONS — Augmented Cell 4.2  (scope & composition)
# ============================================================

# [CORE] Variables defined inside a function are LOCAL — invisible outside.
#        Variables defined outside are GLOBAL — readable inside but not
#        writable without the 'global' keyword (avoid global writes).
#        Functions can call other functions — this is composition.

TAX_RATE = 0.09   # global constant — uppercase signals it's not meant to change

def apply_tax(amount):
    # [CORE] 'amount' and 'tax' are local — they disappear after the function returns
    tax = amount * TAX_RATE    # reading a global is fine
    return round(amount + tax, 2)

def full_receipt(covers, price_per_cover, tip_percent=18):
    # [CORE] Composition: call calculate_bill, then pipe result into apply_tax
    bill = calculate_bill(covers, price_per_cover, tip_percent)
    taxed_total = apply_tax(bill["total"])
    return {
        **bill,                          # unpack all keys from bill dict
        "tax": round(taxed_total - bill["total"], 2),
        "grand_total": taxed_total
    }

receipt = full_receipt(4, 47.50)
for key, value in receipt.items():
    print(f"  {key:>15}: ${value:.2f}")

# [ORBIT: Restaurant] Scope is the KITCHEN BOUNDARY. What happens in the
#                     prep station stays in the prep station. The sauce
#                     reduced in prep (local variable) is handed to the
#                     line cook as a finished product (return value) —
#                     the line cook doesn't reach into prep to grab it raw.
#                     TAX_RATE is the posted rate on the wall — every station
#                     can READ it but no one scribbles over it.

# [ORBIT: Construction] Scope is the SUBCONTRACTOR'S SITE OFFICE.
#                       The electrician's measurements (local vars) stay
#                       in their office. They submit a signed report (return
#                       value) to the general contractor. Building code
#                       (global constant) is readable by all; only the
#                       municipality can amend it.

# [MIRROR] In a kitchen you might physically walk into the prep station
#          and grab something without permission. Python's scoping enforces
#          the rule mechanically — there is no 'walking in'. Local variables
#          are literally inaccessible from outside without explicitly returning them.

# [TRAP] The ** (double-star) unpacking in {**bill, ...} is powerful but
#        if bill already has a key you're also adding, the LAST one wins.
#        Order matters in dictionary unpacking. Later keys overwrite earlier ones.
```
